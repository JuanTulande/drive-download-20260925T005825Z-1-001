## **1. Carga de datos**

Proporcione únicamente el código de Python necesario para descargar el conjunto de datos que será utilizado en el proyecto. Si es necesario, realice operaciones de adquisición e integración del conjunto de datos.

Como primer paso cargamos las librerias necesarias tanto para leer los datos como para realizar los análisis posteriores pertinentes

```python
import pandas as pd
import numpy as np
```

Luego cargamos el dataset, donde se le hacen cambios de nombres y tipos para algunas columas

```python
url = "https://github.com/jdizz0414/Data/raw/refs/heads/main/Eventos%20en%20Salud%202025%20en%20Crudo.xlsx"

df = pd.read_excel(url,dtype={"código divipola": str,"código divipola.1": str, "CIE-10": str})
df = df.rename(columns={ "fecha de registro": "FECHA REGISTRO","código divipola": "COD DIVIPOLA DEPARTAMENTO",
                        "departamento": "DEPARTAMENTO","código divipola.1": "COD DIVIPOLA MUN",
                        "municipio": "MUNICIPIO", "municipio": "MUNICIPIO",
                         "sistema": "SISTEMA", "entidad reporta": "ENTIDAD REPORTA"})
df.head()
```

Luego se carga informacion sobre algunas caracteristicas relevantes que ofrece la OMS sobre las diferentes enfermedad, caracteristicas como :


*   Contagiosa:  Si cuando existe un mecanismo reconocido de transmision entre personas: directo (contacto, respiratorio, sexual), indirecto (fecal-oral, sangre y hemoderivados), vertical de madre a hijo, o por vector cuando el ser humano es reservorio del agente
*   Crónica: Si cuando la HISTORIA NATURAL de la condicion, no su tratamiento, es de larga duracion: persiste tipicamente mas de seis meses, es recurrente por definicion, o requiere manejo indefinido.
*   Mortalidad: Cada nivel se anclo a la letalidad esperada de la condicion (case fatality rate) en condiciones de atencion habituales, o a la supervivencia relativa a cinco años cuando se trata de tumores malignos:1 = muy baja: letalidad por debajo de 0,1%, o supervivencia a 5 años superior al 95%, 2 = baja: letalidad entre 0,1% y 1%, o supervivencia a 5 años entre 80% y 95%, 3 = moderada: letalidad entre 1% y 10%, o supervivencia a 5 años entre 50% y 80%,  4 = alta: letalidad entre 10% y 50%, o supervivencia a 5 años entre 20% y 50%, 5 = muy alta: letalidad superior al 50%, o supervivencia a 5 años inferior al 20%

Se quiere cruzar la informacion del dataset original con la obtenida en la OMS con el fin de añadir informacion relevante al dataset


```python
url2 = "https://github.com/jdizz0414/Data/raw/refs/heads/main/Tabla%20auxiliar%20COD%20CIE10.xlsx"

df_aux = pd.read_excel(url2,dtype={"codigo_cie10 	": str,"mortalidad_1_5": str})
df_aux = df_aux.rename(columns={ "codigo_cie10": "CIE-10","enfermedad": "NOMBRE","contagiosa": "CONTAGIOSA",
                         "cronica": "CRONICA", "mortalidad_1_5": "MORTALIDAD"})
df_aux.head()
```

Ahora se hace la union de las dos bases de datos

```python
df_compacto = df.merge(df_aux[["CIE-10","GRUPO","CONTAGIOSA", "CRONICA", "MORTALIDAD"]],on="CIE-10",how="left")
df_compacto.head()
```

### **2.2. Análisis estadístico descriptivo**
---

Describa el conjunto de datos por medio de estadística descriptiva, aportando interpretaciones precisas que ayuden a entender los datos.

```python
columnas_excluir = ["CIE-10", "NOMBRE", "DESCRIPCION"]

for columna in df_compacto.select_dtypes(include="object").columns:

    if columna not in columnas_excluir:

        frecuencia = df_compacto[columna].value_counts()
        proporcion = df_compacto[columna].value_counts(normalize=True) * 100

        tabla = pd.DataFrame({"Frecuencia": frecuencia,"Proporción (%)": proporcion.round(2)})

        tabla.loc["TOTAL"] = [tabla["Frecuencia"].sum(),100.00]

        print("\n", columna)
        print(tabla)
```

AL tratarse de variables categoricas, se hicieron tablas de frecuencia para las vaiables representativas del dataset donde se pueden observar la distribucion de las categorias

- Analice la distribución de las etiquetas, identifique si hay desbalanceo de datos.

Se evidencia un desbalance significativo en la distribución de las categorías de la mayoría de las variables del conjunto de datos. La magnitud de este desbalance varía considerablemente dependiendo de la variable seleccionada como etiqueta objetivo (variable dependiente), lo cual puede tener implicaciones importantes en el desempeño y la interpretación de los modelos de clasificación.

• CONTAGIOSA — Desbalance severo (clasificación binaria)
La variable presenta un desbalance marcado entre sus dos categorías. La clase mayoritaria "No" concentra el 96,44 % de los registros, mientras que la clase minoritaria "Sí" representa únicamente el 3,56 %. Esta distribución corresponde a un desbalance severo, que debe ser considerado al desarrollar modelos de clasificación binaria, ya que un modelo podría favorecer la clase mayoritaria y presentar un desempeño deficiente en la identificación de los casos minoritarios.

• MORTALIDAD — Desbalance ordinal (clasificación multiclase)
La distribución de la variable presenta un sesgo positivo, con una mayor concentración de observaciones en los niveles inferiores de mortalidad. El nivel 1 concentra el 45,98 % de los casos, mientras que el nivel 5 representa únicamente el 1,42 %. En términos relativos, existe una proporción superior a 32:1 entre ambos extremos de la escala, evidenciando una distribución considerablemente desigual entre las categorías.

• GRUPO DIAGNÓSTICO — Desbalance multiclase extremo
Esta variable presenta uno de los niveles de desbalance más elevados del conjunto de datos. La categoría "Enfermedades del aparato circulatorio" concentra el 23,39 % del total de registros. En contraste, las cinco categorías con menor frecuencia, consideradas conjuntamente, no alcanzan el 0,05 % de las observaciones. Adicionalmente, se identifican categorías con una representación prácticamente nula; por ejemplo, "Ciertas afecciones originadas en el período perinatal" presenta una proporción reportada de 0,00 %, aunque cuenta con 17 registros.

• CRÓNICA — Desbalance moderado (clasificación binaria)
La variable presenta un desbalance moderado entre sus categorías. La categoría "Sí" representa el 73,20 % de los registros, mientras que la categoría "No" corresponde al 26,80 %. Aunque la diferencia es considerable, el nivel de desbalance es menor que el observado en variables como CONTAGIOSA.

• DEPARTAMENTO y MUNICIPIO — Concentración y sesgo geográfico
Las variables asociadas a la ubicación geográfica presentan una alta concentración de registros en determinadas zonas del país. El departamento del Valle del Cauca concentra el 32,31 % de los registros, mientras que Cali, perteneciente a este departamento, representa el 22,22 % del total de registros correspondientes a municipios. En contraste, algunos departamentos presentan una representación extremadamente baja; por ejemplo, La Guajira cuenta con menos de tres registros.
