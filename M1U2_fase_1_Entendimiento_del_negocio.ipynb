¿Quienes son los beneficiarios del proyecto? ¿En qué dominio se encuentran (marketing, medicina, entre otros)?
--
*Beneficiarios primarios:*
El Ministerio de Salud y Protección Social, como responsable de la Hoja de Ruta de Datos Abiertos Estratégicos del sector, y las secretarías de salud departamentales y municipales, que necesitan evidencia territorial para priorizar programas de enfermedades crónicas no transmisibles (ENT).

*Beneficiarios secundarios:* las Entidades Administradoras de Planes de Beneficios (EAPB), que reportan los eventos vía RIPS y pueden usar el análisis para compararse frente a otras entidades, y las IPS de su red de prestación.

*Beneficiarios terciarios:* ciudadanía, academia, periodistas y desarrolladores que reutilizan el dato publicado en datos.gov.co, en línea con los usuarios previstos por el Plan de Datos Abiertos de MinSalud.

*Beneficiario interno:* el propio equipo de gobernanza de datos, para quien el producto funciona como control de calidad antes de cada ciclo de publicación.

El proyecto se ubica en el dominio de la salud pública y la epidemiología, con una capa transversal de gobernanza y analítica de datos (no es un proyecto de marketing ni de negocio comercial): combina conocimiento clínico-administrativo (taxonomía CIE-10, RIPS) con un marco de gobernanza de datos aplicado a un conjunto de datos abiertos de cumplimiento normativo.

¿Cuáles son los principales desafíos o problemas que motivan el proyecto?
--
*   Colombia exige a las entidades del sector salud publicar datos abiertos estratégicos conforme a la Hoja de Ruta liderada por MinTIC (estrategia “Transfórmate con Gobierno Digital”) y operacionalizada por MinSalud mediante su Plan de Datos Abiertos, pero no existe hoy un mecanismo sistemático para verificar, antes de publicar, si un extracto como este es apto para reutilización (fit for use).
*   Necesidad real de priorización territorial de recursos frente a la carga de enfermedades crónicas no transmisibles y si esa priorización es confiable en terminos de integridad y gobernabilidad.

---

- ¿Cuáles son los límites y fronteras del proyecto de análisis de datos? ¿Qué aspectos específicos del negocio se incluirán y cuáles se excluyen en el alcance del proyecto?

¿Qué solución se desea implementar?
--
*   Un panel/informe analítico de priorización territorial de ENT a partir del corte RIPS 2025 (departamento, municipio, diagnóstico CIE-10, volumen y tendencia de eventos).
*   Una ficha o tablero de calidad de datos (data quality scorecard) que evalúe el conjunto frente a dimensiones de gobernanza —completitud de cobertura territorial, diversidad de fuentes, consistencia de metadatos y trazabilidad del proceso de extracción— y sirva de insumo para el siguiente ciclo de publicación en cumplimiento de la Hoja de Ruta.

¿Cuáles son los límites del proyecto? ¿Qué se incluye y qué se excluye?
--
Incluye:
--
*   Análisis exclusivo del “Eventos en Salud 2025” (Fuente: RIPS, reportado por EAPB), tal como fue entregado.
*   Agregación geográfica a nivel de departamento y municipio (DIVIPOLA) y por evento CIE-10.
*   Evaluación de calidad de datos del propio archivo (perfilamiento cuantitativo y dimensiones de gobernanza)
*   Recomendaciones de gobernanza para el próximo ciclo de publicación abierta.

Excluye:
--
*   Datos clínicos identificables o a nivel de paciente (el dataset no los contiene: no hay identificador de paciente, edad ni sexo).
*   Otras fuentes de vigilancia como SIVIGILA o INS, contempladas en el diccionario de datos pero ausentes en este corte.
*   Modelos de predicción diagnóstica individual o cualquier uso que requiera reidentificación de personas.
*   Intervención directa sobre los sistemas fuente (RIPS/EAPB) o sobre el proceso ETL interno de MinSalud/ADRES.

¿De qué forma el beneficiario utilizará el producto del proyecto?
--
Se propone que las secretarías de salud y el equipo de gobernanza usen el panel de priorización como insumo de apoyo a la decisión (no como decisión automática) dentro de sus comités de planeación territorial, y usen la ficha de calidad como lista de verificación previa a cada publicación en datos.gov.co, dentro del ciclo institucional “identificación → análisis → priorización → documentación-publicación → programación” que define el Plan de Datos Abiertos de MinSalud.

### **2.1. Origen**
---

- ¿De dónde vienen los datos?
- ¿Se usa alguna herramienta o proceso para la descarga de la información?

¿De dónde vienen los datos?
--
El archivo consolida Registros Individuales de Prestación de Servicios de Salud (RIPS), reportados por las Entidades Administradoras de Planes de Beneficios (EAPB) al Ministerio de Salud y Protección Social, en cumplimiento de la Hoja de Ruta de Datos Abiertos Estratégicos liderada por MinTIC bajo la estrategia “Transfórmate con Gobierno Digital” y operacionalizada para el sector salud mediante el Plan de Datos Abiertos de MinSalud. Ese plan define un ciclo institucional de publicación —identificación, análisis, priorización, documentación-publicación y programación— que culmina en la disposición del dato en el portal datos.gov.co.

¿Se usa alguna herramienta o proceso para la descarga de la información?
--
Actualmente no se tiene alguna herramienta o proceso para la descarga. Los RIPS son enviados por el prestador en excel de manera mensual.

¿Qué tipo de datos está manejando?
--
*   Datos administrativos-epidemiológicos secundarios (no primarios): eventos de atención en salud codificados con CIE-10, no historia clínica completa.
*   Datos estructurados y tabulares: 415.154 registros y 10 columnas, distribuidos en dos hojas (registros + diccionario de datos).
*   Variables: fecha de registro (fecha-hora), ubicación geográfica DIVIPOLA (departamento y municipio, con su código), diagnóstico CIE-10 (código, nombre, descripción), sistema de información de origen y entidad reportante.
*   Datos agregados a nivel de evento, no de persona: no contiene identificador de paciente ni variables demográficas (edad, sexo, régimen de afiliación), por lo que corresponde a un dato abierto ya anonimizado/agregado.
